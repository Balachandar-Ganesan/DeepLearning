```python
import torch
from transformers import pipeline

# Choose a model checkpoint, for example a smaller variant for easier use:
# "bigscience/bloom-1b1" is a 1.1B parameter model
# "bigscience/bloomz-7b1" is an instruction-tuned model better at following prompts
checkpoint = "bigscience/bloomz-7b1"

# Load the pipeline, utilizing auto device mapping for efficient use of GPU/CPU
generator = pipeline(
    'text-generation',
    model=checkpoint,
    torch_dtype="auto",
    device_map="auto"
)


```

```python
# Define your prompt
prompt = "Most important algorithms for searching  are"

# Generate text
outputs = generator(
    prompt,
    max_length=50,      # Max length of the generated sequence
    num_return_sequences=1, # Number of different sequences to generate
    temperature=0.7     # Controls the randomness of the generation (lower is more deterministic)
)

# Print the generated text
print(outputs[0]['generated_text'])
```
