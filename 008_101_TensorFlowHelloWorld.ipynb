```python
import tensorflow as tf
import numpy as np

# 1. Generate Dummy Dataset (20 records)
np.random.seed(42)
S1 = np.random.uniform(5, 10, 20)
S2 = np.random.uniform(5, 10, 20)
S3 = np.random.uniform(5, 10, 20) # Present in input but not in formula

# Formula: ((4*S1) + (3*S2) + (2*S1)) / 9.0
CGPA = ((4 * S1) + (3 * S2) + (2 * S1)) / 9.0

X = np.stack([S1, S2, S3], axis=1)
y = CGPA.reshape(-1, 1)

# 2. Build the ANN Model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[3])
])

model.compile(optimizer=tf.keras.optimizers.Adam(0.1), loss='mean_squared_error')

# 3. Train the Model
print("Training the model...")
model.fit(X, y, epochs=500, verbose=0)

# 4. Test the Model with (5, 5, 5)
test_input = np.array([[5.0, 5.0, 5.0]])
prediction = model.predict(test_input)
actual = ((4*5) + (3*5) + (2*5)) / 9.0

print(f"\nTest Input: S1=5, S2=5, S3=5")
print(f"Predicted CGPA: {prediction[0][0]:.4f}")
print(f"Actual Formula CGPA: {actual:.4f}")

# 5. Print Model Parameters (Weights and Bias)
weights, bias = model.layers[0].get_weights()
print("\n--- Model Parameters ---")
print(f"Weights (w1, w2, w3): {weights.flatten()}")
print(f"Bias (b): {bias[0]}")
```

```python
def manual_cgpa_predictor(s1, s2, s3, model):
    """
    Calculates CGPA using the weights and bias from the trained TF model.
    Formula: CGPA = (w1*s1 + w2*s2 + w3*s3) + bias
    """
    weights, bias = model.layers[0].get_weights()
    w1, w2, w3 = weights.flatten()
    b = bias[0]

    calculated_cgpa = (w1 * s1) + (w2 * s2) + (w3 * s3) + b
    return calculated_cgpa

# Example usage:
result = manual_cgpa_predictor(5, 5, 5, model)
print(f"\nFunction Output for (5,5,5): {result:.4f}")
```
