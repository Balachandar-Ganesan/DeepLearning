```python
import tensorflow as tf

# In TF 2.x, Sessions and manual graph resets are no longer needed.
# Everything executes immediately (Eager Execution).

# 1. Create Tensors
my_tensor = tf.zeros([1, 20])

# 2. Declare Variables
# Variables no longer need a manual sess.run(initializer)
my_var = tf.Variable(tf.zeros([1, 20]))

row_dim, col_dim = 2, 3

# Zero and One initialized variables
zero_var = tf.Variable(tf.zeros([row_dim, col_dim]))
ones_var = tf.Variable(tf.ones([row_dim, col_dim]))

# 3. Shaped like other variables
# No need to call sess.run() before using these
zero_similar = tf.Variable(tf.zeros_like(zero_var))
ones_similar = tf.Variable(tf.ones_like(ones_var))

# 4. Filling with constants
fill_var = tf.Variable(tf.fill([row_dim, col_dim], -1))
const_var = tf.Variable(tf.constant([8, 6, 7, 5, 3, 0, 9]))
const_fill_var = tf.Variable(tf.constant(-1, shape=[row_dim, col_dim]))

# 5. Sequence generation
linear_var = tf.Variable(tf.linspace(0.0, 1.0, 3))
sequence_var = tf.Variable(tf.range(start=6, limit=15, delta=3))

# 6. Random Numbers
# 'random_normal' is now 'random.normal'
rnorm_var = tf.random.normal([row_dim, col_dim], mean=0.0, stddev=1.0)

# 7. TensorBoard Summaries
# TF 2.x uses a different API for logging (tf.summary.create_file_writer)
writer = tf.summary.create_file_writer("/tmp/variable_logs")

with writer.as_default():
    # Example: logging the random tensor
    tf.summary.histogram("random_normal", rnorm_var, step=0)
    writer.flush()

# To view the values, simply print them:
print("Sequence Variable:", sequence_var.numpy())
```
